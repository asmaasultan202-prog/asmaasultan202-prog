```python
import math

def area_of_circle(r):
    return math.pi * r * r

def perimeter_of_circle(r):
    return 2 * math.pi * r
```

```python
def add_all_nums(*args):
    total = 0
    for num in args:
        if isinstance(num, (int, float)):
            total += num
        else:
            return "All items must be numbers"
    return total
```

```python
def convert_celsius_2_fahrenheit(celsius):
    return (celsius * 9/5) + 32
```

```python
def check_season(month):
    month = month.capitalize()
    if month in ['September', 'October', 'November']:
        return 'Autumn'
    elif month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    else:
        return 'Invalid Month'
```

```python
def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 == 0:
        return "Slope is undefined"
    return (y2 - y1) / (x2 - x1)
```

```python
def print_list(lst):
    for item in lst:
        print(item)
```

```python
def capitalize_list_items(lst):
    return [str(item).capitalize() for item in lst]
```

```python
def add_item(lst, item):
    new_lst = lst.copy()
    new_lst.append(item)
    return new_lst
```

```python
def remove_item(lst, item):
    new_lst = lst.copy()
    if item in new_lst:
        new_lst.remove(item)
    return new_lst

# استدعاء الدالة لطباعة الناتج:
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))
```

```python
def sum_of_numbers(n):
    return sum(range(1, n + 1))

sum_all_numbers = sum_of_numbers

# استدعاء الدالة لطباعة الناتج:
print(sum_of_numbers(5))
print(sum_all_numbers(10))
print(sum_all_numbers(100))
```

```python
def sum_of_odds(n):
    return sum(i for i in range(1, n + 1) if i % 2 != 0)
```

```python
def sum_of_even(n):
    return sum(i for i in range(1, n + 1) if i % 2 == 0)
```
