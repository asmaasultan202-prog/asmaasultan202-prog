````python
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[2])

try:
    my_tuple[1] = 10
except TypeError as error:
    print(error)

print(my_tuple)
````

````python
my_tuple = ('apple', 'banana', 'cherry')
print(my_tuple[-1])
print('banana' in my_tuple)
print(my_tuple)
````

````python
person = {'name': 'John', 'age': 25, 'city': 'New York'}
print(person['name'])

person['age'] = 26
person['job'] = 'Engineer'
print(person)
````
